## 1. Criterion Definitions

| **Fluency** | Sentences flow naturally — no awkward phrasing, abrupt transitions, or robotic repetition. ## 2. Pass / Fail Definition

### 2a. Cumulative Pass Bar

I chose the following description which **passes** if ALL of the following are met:

- **Min good:** At least **3 out of 7** criteria are rated *good*.
- **Max ok:** At most **3** criteria are rated *ok*.
- **Max bad:** **Zero** criteria are rated *bad*.

Rationale: requiring 3 good ratings ensures enough quality dimensions are genuinely strong. Allowing up to 3 ok ratings accommodates descriptions that are solid across most criteria without being exceptional in all of them.

### 2b. Go / No-Go Rules (Automatic Failure)

The following three criteria trigger an **immediate FAIL** regardless of all other scores:

| Trigger | Rationale |
|---------|----------|
| **Grounding = bad** | Hallucinated or contradicted product facts are a legal and brand trust risk — unpublishable under any circumstances. |
| **Grammar = bad** | Three or more errors, or any meaning-altering error, makes copy unpublishable and damages brand credibility. |
| **Length = bad** | A description below 40 words fails to communicate product value; above 110 words exceeds the stated business requirement. |

```python
PASS_BAR = {
    "min_good": 3,
    "max_ok":   3,    # anything beyond 3 "ok" is a fail even with zero "bad"
    "max_bad":  0,    # any single "bad" is an automatic fail (see go/no-go below)
}

GO_NOGO_RULES = {
    "grounding": ["bad"],          # hallucinated facts → always reject
    "grammar":   ["bad"],          # broken grammar   → always reject
    "length":    ["bad"],          # too short/long   → always reject
}
```

## 3. Scoring Function

`score_description(ratings)` accepts a dict mapping each criterion name to its verdict (`'good'`, `'ok'`, or `'bad'`) and returns `'pass'` or `'fail'`.

The logic runs in two steps:
1. **Go / no-go check** (fast exit) - if any critical criterion is `'bad'`, return `'fail'` immediately.
2. **Cumulative bar check** - count verdicts across all seven criteria and apply the pass bar.

```python
def score_description(ratings: dict) -> str:
    """
    Apply the rubric to a dict of {criterion: verdict} and return 'pass' or 'fail'.

    Parameters
    ----------
    ratings : dict
        Keys are criterion names (must match RUBRIC keys).
        Values are 'good', 'ok', or 'bad'.

    Returns
    -------
    'pass' or 'fail'
    """
    # Step 1 — go / no-go check
    for criterion, failing_verdicts in GO_NOGO_RULES.items():
        if ratings.get(criterion) in failing_verdicts:
            return "fail"

    # Step 2 — cumulative bar
    counts = {"good": 0, "ok": 0, "bad": 0}
    for verdict in ratings.values():
        counts[verdict] = counts.get(verdict, 0) + 1

    if (
        counts["good"] >= PASS_BAR["min_good"]
        and counts["ok"]  <= PASS_BAR["max_ok"]
        and counts["bad"] == 0
    ):
        return "pass"

    return "fail"


def get_test_cases() -> list:
    return [
        # All good → pass
        ({"fluency": "good", "grammar": "good", "tone": "good",
          "length": "good", "grounding": "good", "latency": "good", "cost": "good"},
         "pass"),
        # One bad in grounding → fail (go/no-go)
        ({"fluency": "good", "grammar": "good", "tone": "good",
          "length": "good", "grounding": "bad", "latency": "good", "cost": "good"},
         "fail"),
        # Only 2 good, 5 ok, zero bad → fail (fewer than 3 good)
        ({"fluency": "ok", "grammar": "ok", "tone": "ok",
          "length": "good", "grounding": "good", "latency": "ok", "cost": "ok"},
         "fail"),
        # Exactly 3 good, 3 ok, 0 bad → pass
        ({"fluency": "good", "grammar": "good", "tone": "good",
          "length": "ok", "grounding": "ok", "latency": "ok", "cost": "good"},
         "pass"),
    ]
```

## 4. Self-Test

```python
print("Self-test results:")
for ratings, expected in get_test_cases():
    result = score_description(ratings)
    status = "OK" if result == expected else "FAIL"
    print(f"  [{status}] expected={expected}, got={result}  ratings={ratings}")
```
